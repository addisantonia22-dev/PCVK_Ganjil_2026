```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow #for image diplasy
from skimage import io
from skimage import transform
from matplotlib import pyplot as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls =["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
      "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
    "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTxuEkeNS2MoMZ2F1XX033dnvmQv-GQyW6nc6xWLOhZbw3LhCVeCBLgRziJ&s=10"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2. shape[1]
print("resolusi image: tinggi x lebar = ", tinggi," x ", lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 =cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

TUGAS PRATIKUM D2

```python
# ===== Garis Vertikal di Tengah Image =====
image_4 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_5 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis vertikal ditengah image
for x in range(tinggi):
  image_5[x, int(lebar/2)] = [255, 255, 255]

final_frame_vertikal = cv.hconcat((image_4, image_5))
print("Garis Vertikal:")
cv2_imshow(final_frame_vertikal)
print('\n')
```

```python
#ambil ukuran gambar terlebih dahulu
tinggi, lebar, _ = image.shape

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
#membuat garis menyilang (Diagonal)
for i in range(tinggi):
    for j in range(lebar):
        #garis diagonal dari kiri atas ke kanan bawah
        if i == int(j * tinggi / lebar):
            image_3[i, j] = [255, 255, 255]

        #garis diagonal dari kanan atas ke kiri bawah
        if i == int((lebar - j - 1) * tinggi / lebar):
            image_3[i, j] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_8 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_9 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# tentukan panjang garis (dalam pixel), silakan diubah sesuai kebutuhan
panjang_garis = int(lebar * 0.3)  # contoh: 30% dari lebar image

# supaya garis berada di tengah secara horizontal
mulai_x = int((lebar - panjang_garis) / 2)
akhir_x = mulai_x + panjang_garis

for y in range(mulai_x, akhir_x):
  image_9[int(tinggi/2), y] = [255, 255, 255]

final_frame_panjang = cv.hconcat((image_8, image_9))
print(f"Garis Horizontal dengan panjang {panjang_garis} pixel:")
cv2_imshow(final_frame_panjang)
```

```python
image_10 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_11 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# tentukan area kotak (silakan sesuaikan koordinatnya)
kotak_awal_y = int(tinggi * 0.5)
kotak_akhir_y = int(tinggi * 0.8)
kotak_awal_x = int(lebar * 0.2)
kotak_akhir_x = int(lebar * 0.5)

# mengubah kumpulan pixel pada area kotak menjadi warna putih
image_11[kotak_awal_y:kotak_akhir_y, kotak_awal_x:kotak_akhir_x] = [255, 255, 255]

final_frame_kotak = cv.hconcat((image_10, image_11))
print("Kotak Putih pada Image:")
cv2_imshow(final_frame_kotak)
```
