```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

print('Mengubah tingkat kecerahan citra dengan Transformasi Log')
print('------------------------------------------------------')

try:
    brightness = float(input('Masukkan nilai kecerahan (konstanta c): '))
except ValueError:
    print('Error, masukkan angka yang valid')
    brightness = 1.0

folder_path = '/content/drive/MyDrive/PCVK/'

original_1 = cv2.imread(folder_path + 'houses.jpg')
original_2 = cv2.imread(folder_path + 'dani.jpg')

if original_1 is None or original_2 is None:
    print("Error: Gambar tidak ditemukan! Periksa kembali nama file Anda.")
else:
    log_image_1 = brightness * np.log1p(original_1.astype(np.float64))
    log_image_1 = np.clip(log_image_1, 0, 255).astype(np.uint8)

    log_image_2 = brightness * np.log1p(original_2.astype(np.float64))
    log_image_2 = np.clip(log_image_2, 0, 255).astype(np.uint8)

    h1, w1 = original_1.shape[:2]
    original_2_resized = cv2.resize(original_2, (w1, h1))
    log_image_2_resized = cv2.resize(log_image_2, (w1, h1))

    row1 = cv2.hconcat([original_1, log_image_1])
    row2 = cv2.hconcat([original_2_resized, log_image_2_resized])

    final_frame = cv2.vconcat([row1, row2])

    cv2_imshow(final_frame)
```

```python
import matplotlib.pyplot as plt

img_bgr = cv2.imread('/content/drive/MyDrive/PCVK/peppers.jpg')

if img_bgr is None:
    print("Error: Gambar tidak ditemukan! Periksa kembali nama file Anda.")
else:
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    R = img_rgb[:, :, 0].astype(np.float32)
    G = img_rgb[:, :, 1].astype(np.float32)
    B = img_rgb[:, :, 2].astype(np.float32)

    # -------------------------------------------------------------
    # a. Metode Averaging
    # Formula: (R + G + B) / 3
    # -------------------------------------------------------------
    gray_averaging = (R + G + B) / 3.0
    gray_averaging = gray_averaging.astype(np.uint8)

    # -------------------------------------------------------------
    # b. Metode Lightness
    # Formula: (max(R, G, B) + min(R, G, B)) / 2
    # -------------------------------------------------------------
    max_rgb = np.maximum(np.maximum(R, G), B)
    min_rgb = np.minimum(np.minimum(R, G), B)
    gray_lightness = (max_rgb + min_rgb) / 2.0
    gray_lightness = gray_lightness.astype(np.uint8)

    # -------------------------------------------------------------
    # c. Metode Luminance
    # Formula standar: 0.299*R + 0.587*G + 0.114*B
    # -------------------------------------------------------------
    gray_luminance = (0.299 * R) + (0.587 * G) + (0.114 * B)
    gray_luminance = gray_luminance.astype(np.uint8)

    # =============================================================
    # MENAMPILKAN HASIL MENGGUNAKAN MATPLOTLIB
    # =============================================================

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.subplot(1, 2, 2)
    plt.imshow(gray_averaging, cmap='gray')
    plt.title("a. Averaging")
    plt.show()

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.subplot(1, 2, 2)
    plt.imshow(gray_lightness, cmap='gray')
    plt.title("b. Lightness")
    plt.show()

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.subplot(1, 2, 2)
    plt.imshow(gray_luminance, cmap='gray')
    plt.title("c. Luminance")
    plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

folder_path = '/content/drive/MyDrive/PCVK/'

img_bgr1 = cv.imread(folder_path + 'peppers.jpg')
img_bgr2 = cv.imread(folder_path + 'dani.jpg')

if img_bgr1 is None or img_bgr2 is None:
    print("Error: Gambar tidak ditemukan! Periksa kembali nama file Anda.")
else:
    def process_red_splash(img_bgr):
        img_hsv = cv.cvtColor(img_bgr, cv.COLOR_BGR2HSV)

        lower_red1 = np.array([0, 50, 50])
        upper_red1 = np.array([10, 255, 255])
        lower_red2 = np.array([170, 50, 50])
        upper_red2 = np.array([180, 255, 255])

        mask1 = cv.inRange(img_hsv, lower_red1, upper_red1)
        mask2 = cv.inRange(img_hsv, lower_red2, upper_red2)
        mask_red = mask1 | mask2

        gray = cv.cvtColor(img_bgr, cv.COLOR_BGR2GRAY)
        gray_3ch = cv.cvtColor(gray, cv.COLOR_GRAY2BGR)

        return np.where(mask_red[:, :, None] == 255, img_bgr, gray_3ch)

    result1 = process_red_splash(img_bgr1)
    result2 = process_red_splash(img_bgr2)

    plt.figure(figsize=(10, 8))

    plt.subplot(2, 2, 1)
    plt.imshow(cv.cvtColor(img_bgr1, cv.COLOR_BGR2RGB))
    plt.title('Gambar 1 Asli')
    plt.axis('off')

    plt.subplot(2, 2, 2)
    plt.imshow(cv.cvtColor(result1, cv.COLOR_BGR2RGB))
    plt.title('Gambar 1 Splash Merah')
    plt.axis('off')

    plt.subplot(2, 2, 3)
    plt.imshow(cv.cvtColor(img_bgr2, cv.COLOR_BGR2RGB))
    plt.title('Gambar 2 Asli')
    plt.axis('off')

    plt.subplot(2, 2, 4)
    plt.imshow(cv.cvtColor(result2, cv.COLOR_BGR2RGB))
    plt.title('Gambar 2 Splash Merah')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(' Gamma Correction pada citra ')
print('-----------------------------')
try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

folder_path = '/content/drive/MyDrive/PCVK/'

img_bgr1 = cv2.imread(folder_path + 'houses.jpg')
img_bgr2 = cv2.imread(folder_path + 'dani.jpg')

if img_bgr1 is None or img_bgr2 is None:
    print("Error: Gambar tidak ditemukan! Periksa kembali nama file Anda.")
else:

    img_rgb1 = cv2.cvtColor(img_bgr1, cv2.COLOR_BGR2RGB)
    img_rgb2 = cv2.cvtColor(img_bgr2, cv2.COLOR_BGR2RGB)
    def apply_gamma(image, gamma_value):
        normalized = image.astype(np.float32) / 255.0
        corrected = np.power(normalized, gamma_value) * 255.0
        return np.clip(corrected, 0, 255).astype(np.uint8)

    gamma_image1 = apply_gamma(img_rgb1, gamma)
    gamma_image2 = apply_gamma(img_rgb2, gamma)

    plt.figure(figsize=(12, 10))

    plt.subplot(2, 2, 1)
    plt.imshow(img_rgb1)
    plt.title("Gambar 1 Asli")
    plt.axis('off')

    plt.subplot(2, 2, 2)
    plt.imshow(gamma_image1)
    plt.title(f"Gambar 1 Gamma ($\gamma$ = {gamma})")
    plt.axis('off')

    plt.subplot(2, 2, 3)
    plt.imshow(img_rgb2)
    plt.title("Gambar 2 Asli")
    plt.axis('off')

    plt.subplot(2, 2, 4)
    plt.imshow(gamma_image2)
    plt.title(f"Gambar 2 Gamma ($\gamma$ = {gamma})")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

try:
    bit_depth = int(input('Masukkan bit depth tujuan (1-8): '))
except ValueError:
    print('Error, masukkan angka yang valid')
    bit_depth = 2

level = 255 / (pow(2, bit_depth) - 1)

print(f"Bit depth awal   : 8 bit")
print(f"Bit depth tujuan : {bit_depth} bit")
print(f"Jumlah level     : {int(pow(2, bit_depth))}")

folder_path = '/content/drive/MyDrive/PCVK/'

original_1 = cv.imread(folder_path + 'peppers.jpg', cv.IMREAD_GRAYSCALE)
original_2 = cv.imread(folder_path + 'dani.jpg', cv.IMREAD_GRAYSCALE)

if original_1 is None or original_2 is None:
    print("Error: Gambar tidak ditemukan! Periksa kembali nama file Anda.")
else:

    def quantize_image(img, lvl):
        quantized = np.round(img / lvl) * lvl
        return np.clip(quantized, 0, 255).astype(np.uint8)

    depth_image_1 = quantize_image(original_1, level)
    depth_image_2 = quantize_image(original_2, level)

    plt.figure(figsize=(10, 8))

    plt.subplot(2, 2, 1)
    plt.imshow(original_1, cmap='gray')
    plt.title("Gambar 1 Grayscale 8-bit")
    plt.axis('off')

    plt.subplot(2, 2, 2)
    plt.imshow(depth_image_1, cmap='gray')
    plt.title(f"Gambar 1 Grayscale {bit_depth}-bit")
    plt.axis('off')

    plt.subplot(2, 2, 3)
    plt.imshow(original_2, cmap='gray')
    plt.title("Gambar 2 Grayscale 8-bit")
    plt.axis('off')

    plt.subplot(2, 2, 4)
    plt.imshow(depth_image_2, cmap='gray')
    plt.title(f"Gambar 2 Grayscale {bit_depth}-bit")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2 as cv
import numpy as np
import glob
from math import log10, sqrt
from google.colab.patches import cv2_imshow

# ==========================================
# 1. Fungsi Menghitung PSNR
# ==========================================
def PSNR(original, denoised):
    mse = np.mean(
        (original.astype(np.float64) - denoised.astype(np.float64)) ** 2
    )

    if mse == 0:
        return 100

    max_pixel = 255.0
    psnr = 20 * log10(max_pixel / sqrt(mse))

    return psnr


# ==========================================
# 2. Membaca Citra Asli
# ==========================================
path_asli = '/content/drive/MyDrive/PCVK/galaxy.jpg'

img_asli = cv.imread(path_asli)

if img_asli is None:
    print("Citra asli tidak ditemukan!")
else:
    print("Citra asli berhasil dibaca.")
    print("Ukuran citra asli :", img_asli.shape)


# ==========================================
# 3. Membaca Semua Citra Noise
# ==========================================
path_noise = '/content/drive/MyDrive/PCVK/noises/*.jpg'

cv_img = []

for img in sorted(glob.glob(path_noise)):

    n = cv.imread(img)

    if n is not None:

        n = cv.resize(
            n,
            (img_asli.shape[1], img_asli.shape[0])
        )

        cv_img.append(n)


print("\nJumlah citra noise yang terbaca :", len(cv_img))


# ==========================================
# 4. Cek Jumlah Citra
# ==========================================
if len(cv_img) < 100:

    print("PERINGATAN!")
    print("Jumlah citra noise kurang dari 100.")
    print("Jumlah citra yang tersedia :", len(cv_img))

else:

    print("100 citra noise berhasil dibaca.")


# ==========================================
# 5. Menampilkan Citra Asli
# ==========================================
print("\n====================================")
print("CITRA ASLI")
print("====================================")

cv2_imshow(img_asli)


# ==========================================
# 6. Average Denoising
# ==========================================
jumlah_data = [10, 20, 40, 80, 100]

print("\n====================================")
print("HASIL AVERAGE DENOISING DAN PSNR")
print("====================================")


for jumlah in jumlah_data:

    if jumlah > len(cv_img):

        print(
            f"\nJumlah Citra = {jumlah} "
            f"tidak dapat diproses karena hanya "
            f"{len(cv_img)} citra yang tersedia."
        )

        continue


    # --------------------------------------
    # Membuat array kosong
    # --------------------------------------
    avg = np.zeros_like(
        cv_img[0],
        dtype=np.float64
    )


    # --------------------------------------
    # Menjumlahkan citra
    # --------------------------------------
    for i in range(jumlah):

        avg += cv_img[i].astype(np.float64)


    # --------------------------------------
    # Menghitung rata-rata
    # --------------------------------------
    avg = avg / jumlah


    # --------------------------------------
    # Mengubah kembali ke uint8
    # --------------------------------------
    avg = np.clip(avg, 0, 255).astype(np.uint8)


    # --------------------------------------
    # Menghitung PSNR
    # --------------------------------------
    nilai_psnr = PSNR(
        img_asli,
        avg
    )


    # --------------------------------------
    # Menampilkan hasil
    # --------------------------------------
    print(f"\nJumlah Citra : {jumlah}")
    print(f"PSNR         : {nilai_psnr:.2f} dB")

    cv2_imshow(avg)


# ==========================================
# 7. Menyimpan Hasil Average Denoising 100
# ==========================================
if len(cv_img) >= 100:

    hasil_100 = np.zeros_like(
        cv_img[0],
        dtype=np.float64
    )


    for i in range(100):

        hasil_100 += cv_img[i].astype(np.float64)


    hasil_100 = hasil_100 / 100

    hasil_100 = np.clip(
        hasil_100,
        0,
        255
    ).astype(np.uint8)

    output_path = '/content/average_denoising_100.jpg'

    cv.imwrite(
        output_path,
        hasil_100
    )

    print("\n====================================")
    print("HASIL 100 CITRA")
    print("====================================")

    print("PSNR 100 citra : {:.2f} dB".format(
        PSNR(img_asli, hasil_100)
    ))

    print("Hasil disimpan di:")
    print(output_path)
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_bgr = cv.imread('/content/drive/MyDrive/PCVK/couple.tiff')

if img_bgr is not None:
    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
    h, w, _ = img_bgr.shape

    mask = np.zeros((h, w), dtype=np.uint8)
    cv.circle(mask, (int(w * 0.25), int(h * 0.22)), 45, 255, -1)
    cv.circle(mask, (int(w * 0.77), int(h * 0.25)), 45, 255, -1)
    mask_3ch = cv.merge([mask, mask, mask])

    # ==========================================
    # 1. OPERATOR NOT (Komplemen)
    # ==========================================
    res_not = cv.bitwise_not(img_rgb)

    plt.figure(figsize=(8, 4))
    plt.subplot(1, 2, 1)
    plt.title("1. Input Citra")
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title("Output NOT (Komplemen)")
    plt.imshow(res_not)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    # ==========================================
    # 2. OPERATOR OR (Atau)
    # ==========================================
    res_or = cv.bitwise_or(img_rgb, mask_3ch)

    plt.figure(figsize=(10, 4))
    plt.subplot(1, 3, 1)
    plt.title("2. Input Citra")
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.title("Input Mask")
    plt.imshow(mask_3ch)
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.title("Output OR")
    plt.imshow(res_or)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    # ==========================================
    # 3. OPERATOR AND (Dan)
    # ==========================================
    res_and = cv.bitwise_and(img_rgb, mask_3ch)

    plt.figure(figsize=(10, 4))
    plt.subplot(1, 3, 1)
    plt.title("3. Input Citra")
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.title("Input Mask")
    plt.imshow(mask_3ch)
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.title("Output AND")
    plt.imshow(res_and)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    # ==========================================
    # 4. OPERATOR NAND (Not AND)
    # ==========================================
    res_nand = cv.bitwise_not(res_and)

    plt.figure(figsize=(10, 4))
    plt.subplot(1, 3, 1)
    plt.title("4. Input Citra")
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.title("Input Mask")
    plt.imshow(mask_3ch)
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.title("Output NAND")
    plt.imshow(res_nand)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

    # ==========================================
    # 5. OPERATOR XOR (Exclusive OR)
    # ==========================================
    res_xor = cv.bitwise_xor(img_rgb, mask_3ch)

    plt.figure(figsize=(10, 4))
    plt.subplot(1, 3, 1)
    plt.title("5. Input Citra")
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.title("Input Mask")
    plt.imshow(mask_3ch)
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.title("Output XOR")
    plt.imshow(res_xor)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
else:
    print("Gagal memuat citra! Periksa kembali path file couple.tiff Anda.")
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

path_foto_malam = '/content/drive/MyDrive/PCVK/malam.jpg'
img_bgr = cv.imread(path_foto_malam)

if img_bgr is not None:
    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

    gamma = 0.4
    inv_gamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** inv_gamma) * 255 for i in np.arange(0, 256)]).astype("uint8")

    img_gamma = cv.LUT(img_rgb, table)

    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Foto Malam Asli (Gelap)")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(img_gamma)
    plt.title(f"Hasil Gamma Correction (γ = {gamma})")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gagal memuat gambar. Pastikan path file foto malam Anda benar.")
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_bgr = cv.imread('/content/drive/MyDrive/PCVK/crayfish.jpg')

if img_bgr is not None:
    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

    img_stretched = np.zeros_like(img_rgb)
    for i in range(3):
        min_val = np.min(img_rgb[:, :, i])
        max_val = np.max(img_rgb[:, :, i])
        img_stretched[:, :, i] = ((img_rgb[:, :, i] - min_val) / (max_val - min_val) * 255).astype(np.uint8)

    gamma = 1.2
    inv_gamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** inv_gamma) * 255 for i in np.arange(0, 256)]).astype("uint8")
    img_final = cv.LUT(img_stretched, table)

    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Citra Asli (Low Contrast / Hazy)")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(img_final)
    plt.title("Hasil Perbaikan (Stretching + Gamma 1.2)")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gagal memuat citra crayfish.jpg!")
```
